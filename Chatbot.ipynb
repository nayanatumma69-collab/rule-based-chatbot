# 🤖 Rule-Based Chatbot

## 📌 Project Overview

This project implements a simple **rule-based conversational chatbot** using Python, Regular Expressions (Regex), and pattern matching.

The chatbot identifies the user's intent by matching their input against predefined keyword and Regex patterns. Based on the detected intent, it provides a randomly selected response from a predefined response pool.

The chatbot is designed to handle common conversations related to an **AI Internship**, including greetings, internship information, deadlines, technical support, and chatbot capabilities.

---

## 🎯 Objective

The main objective of this project is to understand the basic concepts behind **Conversational AI and Natural Language Processing (NLP)** using a rule-based approach without using machine learning or Large Language Models (LLMs).

## ⚙️ How the Chatbot Works

The chatbot follows these steps:

1. **Takes input** from the user.
2. **Cleans the input** by converting it to lowercase and removing punctuation.
3. **Compares the cleaned input** with predefined Regex patterns.
4. **Identifies the corresponding user intent.**
5. **Selects a response** from the response pool for that intent.
6. **Returns the response** to the user.
7. If no matching intent is found, the chatbot provides a **fallback response**.

````python
import random
import re


class RuleBasedChatbot:

    def __init__(self):
        # Dictionary mapping intents to lists of regular expressions
        self.rules = {
            # Topic 1: Greetings & Small Talk
            "greeting": [
                r"\b(hi|hello|hey|greetings|howdy)\b",
                r"\bgood\s(morning|afternoon|evening)\b",
            ],
            "farewell": [
                r"\b(bye|goodbye|see\syou|exit|quit)\b",
                r"\bhave\sa\sgood\sday\b",
            ],
            # Topic 2: AI Internship & Course Info
            "internship_info": [
                r"\b(internship|program|syllabus|curriculum|duration)\b",
                r"\bwhat\swill\si\slearn\b",
            ],
            "deadline_info": [
                r"\b(deadline|due\sdate|timeline|submission)\b",
                r"\bwhen\sis\sit\sdue\b",
            ],
            # Topic 3: Technical Support
            "tech_support": [
                r"\b(error|bug|issue|failed|crash|broken|not\sworking)\b",
                r"\bhow\sdo\si\sinstall\b",
            ],
            # Topic 4: Bot Capabilities / Identity
            "identity": [
                r"\b(who\sare\syou|what\sare\syou|your\sname)\b",
                r"\bwhat\scan\syou\sdo\b",
            ],
        }

        # Response pools mapped to each detected intent
        self.responses = {
            "greeting": [
                "Hello! How can I assist you with your AI internship today?",
                "Hi there! What can I help you with?",
                "Greetings! Ready to tackle your Week 1 tasks?",
            ],
            "farewell": [
                "Goodbye! Best of luck with your internship assignments.",
                "See you later! Feel free to reach out if you get stuck.",
                "Have a great day ahead! Happy coding!",
            ],
            "internship_info": [
                "The internship covers foundational AI topics including neural networks, NLP, and heuristic algorithms.",
                "Week 1 focuses on AI Foundations: Neural Networks, Rule-Based NLP, or Minimax algorithms.",
            ],
            "deadline_info": [
                "Assignments are typically due at the end of each sprint week. Check your portal for exact timestamps.",
                "Make sure to push your code and report before the weekly review call.",
            ],
            "tech_support": [
                "For environment issues, verify package versions and ensure your runtime is active.",
                "If you face runtime errors, check your Python syntax and code block execution order.",
            ],
            "identity": [
                "I am a rule-based AI assistant built to guide you through your Week 1 curriculum.",
                "I am an automated chatbot that uses regular expressions and pattern matching to answer questions.",
            ],
        }

        # Fallback pool for unmatched inputs
        self.fallbacks = [
            "I'm not quite sure I understand. Could you rephrase that?",
            "Sorry, that is outside my current knowledge base. Try asking about the internship, technical support, or deadlines.",
            "I couldn't match that to a known topic. Can you specify your question with different keywords?",
        ]

    def clean_text(self, text: str) -> str:
        """Standardizes input by lowercasing and stripping punctuation."""
        text = text.lower().strip()
        text = re.sub(r"[^\w\s]", "", text)
        return text

    def get_intent(self, user_input: str) -> str:
        """Scans input against regex patterns to classify user intent."""
        cleaned = self.clean_text(user_input)
        for intent, patterns in self.rules.items():
            for pattern in patterns:
                if re.search(pattern, cleaned):
                    return intent
        return "fallback"
    def respond(self, user_input: str) -> str:
        """Picks a random reply matching the intent, or a fallback reply."""
        intent = self.get_intent(user_input)
        if intent == "fallback":
            return random.choice(self.fallbacks)
        return random.choice(self.responses[intent])

    def run(self):
        """Interactive loop that handles user inputs inside Google Colab."""
        print("=" * 60)
        print("AI Assistant Online (Google Colab Edition)")
        print("Type 'exit', 'quit', or 'bye' to end the conversation.")
        print("=" * 60)

        while True:
            try:
                user_msg = input("\nYou: ")
                if not user_msg.strip():
                    continue

                # Terminate loop if the user says goodbye
                if self.get_intent(user_msg) == "farewell":
                    print(f"Bot: {self.respond(user_msg)}")
                    print("\n[Chat session ended]")
                    break

                bot_msg = self.respond(user_msg)
                print(f"Bot: {bot_msg}")

            except (KeyboardInterrupt, EOFError):
                print("\n[Chat session ended]")
                break


# Run the chatbot
bot = RuleBasedChatbot()
bot.run()
````
